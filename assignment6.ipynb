```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
fahrenheit = corrected_temperature * 9 / 5 + 32
above_32 = corrected_temperature[corrected_temperature > 32]
count_above_32 = np.sum(corrected_temperature > 32)

print("Corrected temperatures:", corrected_temperature)
print("Fahrenheit:", fahrenheit)
print("Readings above 32°C:", above_32)
print("Count above 32°C:", count_above_32)
print("Vectorization applies operations to the entire array efficiently.")
print("Boolean indexing selects values that satisfy a condition without loops.")    



```

```python
#Q2
import numpy as np
steps = np.array([

    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("\nQ2")
print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))
print("Total steps per day:", np.sum(steps, axis=0))
print("Total steps per user:", np.sum(steps, axis=1))

max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("Maximum value position (user, day):", max_position)
print("Maximum value:", steps[max_position])
```

```python
# Q3
import numpy as np
original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
subset[0] = 999

print("\nQ3")
print("Original after modifying slice:", original)
print("Subset:", subset)

copied_subset = original[1:5].copy()
copied_subset[0] = 500

print("Original after modifying copied slice:", original)
print("Copied array:", copied_subset)

matrix = np.arange(1, 13).reshape(3, 4)
print("3 × 4 matrix:\n", matrix)

print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1–2 and columns 2–3:\n", matrix[1:3, 2:4])

flattened = matrix.flatten()
raveled = matrix.ravel()

raveled[0] = 100
print("Matrix after modifying ravel():\n", matrix)
print("Raveled array:", raveled)

flattened[1] = 200
print("Matrix after modifying flatten():\n", matrix)
print("Flattened array:", flattened)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
# Q4
import numpy as np
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("\nQ4")
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

XT = X.T
print("X.T:\n", XT)
print("The transpose changes rows into columns, so each feature becomes a row.")

XTX = XT @ X
print("X.T @ X:\n", XTX)

XTX_inverse = np.linalg.inv(XTX)
print("Inverse of X.T @ X:\n", XTX_inverse)

beta = XTX_inverse @ XT @ y
print("OLS coefficients:", beta)
print("The coefficients represent the contribution of sleep, activity, and stress.")

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("Predicted assistance score:", predicted_score)
```
